```python
# ================================
# 0. Imports
# ================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
sns.set(style="whitegrid", font_scale=1.2)
# Paths
DATA_DIR = Path("../data")
RAW_DIR = DATA_DIR / "raw"
PROC_DIR = DATA_DIR / "processed"

# File paths
TRAIN_PATH = RAW_DIR / "claims_train.csv"
df_train = pd.read_csv(TRAIN_PATH)
# ============================================
# 1. UNIVARIATE ANALYSIS
# ============================================

numeric_cols = ["ClaimNb", "Exposure", "VehPower", "VehAge", "DrivAge", 
                "Density", "BonusMalus"]
categorical_cols = ["VehBrand", "VehGas", "Area", "Region"]

# ---------- 1.1 NUMERIC FEATURES ----------
def univariate_numeric(df, cols):
    for col in cols:
        fig, ax = plt.subplots(1, 2, figsize=(12, 4))

        # Histogram
        sns.histplot(df[col], kde=False, bins=30, ax=ax[0])
        ax[0].set_title(f"Histogram of {col}")

        # Density
        sns.kdeplot(df[col], ax=ax[1])
        ax[1].set_title(f"Density Plot of {col}")

        plt.show()

        # Summary stats
        print(f"\nSummary statistics for {col}:")
        print(df[col].describe())
        print(f"Skewness: {df[col].skew():.3f}")
        print("-" * 60)

univariate_numeric(df_train, numeric_cols)

# ---------- 1.2 CATEGORICAL FEATURES ----------
def univariate_categorical(df, cols):
    for col in cols:
        plt.figure(figsize=(10, 4))
        order = df[col].value_counts().index
        sns.barplot(x=df[col].value_counts().values, y=order)
        plt.title(f"Category Distribution: {col}")
        plt.xlabel("Count")
        plt.ylabel(col)
        plt.show()

        print(df[col].value_counts())
        print("-" * 60)

univariate_categorical(df_train, categorical_cols)

# ============================================
# 2. BIVARIATE ANALYSIS
# ============================================

# ---------- 2.1 NUMERIC vs NUMERIC ----------
pairs_numeric = [
    ("Exposure", "ClaimNb"),
    ("DrivAge", "ClaimNb"),
    ("VehPower", "VehAge"),
    ("Density", "ClaimNb")
]

def scatter_pairs(df, pairs):
    for x, y in pairs:
        plt.figure(figsize=(6, 5))
        sns.scatterplot(x=df[x], y=df[y], alpha=0.3)
        plt.title(f"{x} vs {y}")
        plt.show()

scatter_pairs(df_train, pairs_numeric)

# Correlation matrix
plt.figure(figsize=(10, 6))
corr = df_train[numeric_cols].corr()
sns.heatmap(corr, cmap="coolwarm", annot=True)
plt.title("Correlation Heatmap (Numeric Features)")
plt.show()

# ---------- 2.2 NUMERIC vs CATEGORICAL ----------
def numeric_vs_categorical(df, target, cat_cols):
    for col in cat_cols:
        plt.figure(figsize=(10, 5))
        sns.boxplot(x=col, y=target, data=df)
        plt.title(f"{target} across {col}")
        plt.xticks(rotation=45)
        plt.show()

numeric_vs_categorical(df_train, "ClaimNb", categorical_cols)

# Violin example (BonusMalus)
sns.violinplot(x="Area", y="BonusMalus", data=df_train)
plt.title("BonusMalus distribution across Area")
plt.show()

# ---------- 2.3 CATEGORICAL vs CATEGORICAL ----------
def categorical_contingency(df, col1, col2):
    table = pd.crosstab(df[col1], df[col2])
    print(f"\nContingency Table: {col1} × {col2}")
    display(table)

categorical_contingency(df_train, "Area", "VehGas")
categorical_contingency(df_train, "VehBrand", "Region")

# ============================================
# 3. TARGET VARIABLE ANALYSIS
# ============================================

# ---------- 3.1 ClaimNb distribution ----------
plt.figure(figsize=(10, 4))
sns.histplot(df_train["ClaimNb"], bins=20)
plt.title("Distribution of ClaimNb")
plt.show()

print(df_train["ClaimNb"].value_counts().head(10))

# Zero Inflation
zero_rate = (df_train["ClaimNb"] == 0).mean()
print(f"Zero-Inflation Rate: {zero_rate:.3f}")

# ---------- 3.2 Exposure-adjusted claim rate ----------
df_train["ClaimRate"] = df_train["ClaimNb"] / df_train["Exposure"]

plt.figure(figsize=(10, 4))
sns.histplot(df_train["ClaimRate"], bins=30)
plt.title("Distribution of Exposure-adjusted Claim Rate")
plt.show()

# Check suspicious low exposure values
suspicious = df_train[df_train["Exposure"] < 0.01]
print("Suspicious low exposure rows:")
display(suspicious.head())

# ============================================
# 4. MULTIVARIATE RELATIONSHIPS
# ============================================

# ---------- 4.1 Correlation Analysis ----------
plt.figure(figsize=(10, 6))
sns.heatmap(df_train[numeric_cols].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Structure Among Predictors")
plt.show()

# ---------- 4.2 Interaction Exploration ----------
sns.lmplot(data=df_train, x="DrivAge", y="BonusMalus", hue="Area", scatter_kws={"alpha": 0.2})
plt.title("Interaction: DrivAge × BonusMalus × Area")
plt.show()

sns.boxplot(x="Region", y="ClaimNb", data=df_train)
plt.title("Interaction Example: Region × ClaimNb")
plt.xticks(rotation=45)
plt.show()

```
