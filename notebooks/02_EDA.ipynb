```python
# ================================
# 0. Imports
# ================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from sklearn.preprocessing import StandardScaler

sns.set(style="whitegrid", font_scale=1.2)
# Paths
DATA_DIR = Path("../data")
RAW_DIR = DATA_DIR / "raw"
PROC_DIR = DATA_DIR / "processed"

# File paths
TRAIN_PATH = RAW_DIR / "claims_train.csv"
df_train = pd.read_csv(TRAIN_PATH)
# ============================================
# 1. UNIVARIATE ANALYSIS
# ============================================

numeric_cols = ["ClaimNb", "Exposure", "VehPower", "VehAge", "DrivAge", 
                "Density", "BonusMalus"]
categorical_cols = ["VehBrand", "VehGas", "Area", "Region"]

# ---------- 1.1 NUMERIC FEATURES ----------
def univariate_numeric(df, cols):
    for col in cols:
        fig, ax = plt.subplots(1, 2, figsize=(12, 4))

        # Histogram
        sns.histplot(df[col], kde=False, bins=30, ax=ax[0])
        ax[0].set_title(f"Histogram of {col}")

        # Density
        sns.kdeplot(df[col], ax=ax[1])
        ax[1].set_title(f"Density Plot of {col}")

        plt.show()

        # Summary stats
        print(f"\nSummary statistics for {col}:")
        print(df[col].describe())
        print(f"Skewness: {df[col].skew():.3f}")
        print("-" * 60)

univariate_numeric(df_train, numeric_cols)

# ---------- 1.2 CATEGORICAL FEATURES ----------
def univariate_categorical(df, cols):
    for col in cols:
        plt.figure(figsize=(10, 4))
        order = df[col].value_counts().index
        sns.barplot(x=df[col].value_counts().values, y=order)
        plt.title(f"Category Distribution: {col}")
        plt.xlabel("Count")
        plt.ylabel(col)
        plt.show()

        print(df[col].value_counts())
        print("-" * 60)

univariate_categorical(df_train, categorical_cols)

# ============================================
# 2. BIVARIATE ANALYSIS
# ============================================

# ---------- 2.1 NUMERIC vs NUMERIC ----------
pairs_numeric = [
    ("Exposure", "ClaimNb"),
    ("DrivAge", "ClaimNb"),
    ("VehPower", "VehAge"),
    ("Density", "ClaimNb")
]

def scatter_pairs(df, pairs):
    for x, y in pairs:
        plt.figure(figsize=(6, 5))
        sns.scatterplot(x=df[x], y=df[y], alpha=0.3)
        plt.title(f"{x} vs {y}")
        plt.show()

scatter_pairs(df_train, pairs_numeric)

# Correlation matrix
plt.figure(figsize=(10, 6))
corr = df_train[numeric_cols].corr()
sns.heatmap(corr, cmap="coolwarm", annot=True)
plt.title("Correlation Heatmap (Numeric Features)")
plt.show()

# ---------- 2.2 NUMERIC vs CATEGORICAL ----------
def numeric_vs_categorical(df, target, cat_cols):
    for col in cat_cols:
        plt.figure(figsize=(10, 5))
        sns.boxplot(x=col, y=target, data=df)
        plt.title(f"{target} across {col}")
        plt.xticks(rotation=45)
        plt.show()

numeric_vs_categorical(df_train, "ClaimNb", categorical_cols)

# Violin example (BonusMalus)
sns.violinplot(x="Area", y="BonusMalus", data=df_train)
plt.title("BonusMalus distribution across Area")
plt.show()

# ---------- 2.3 CATEGORICAL vs CATEGORICAL ----------
def categorical_contingency(df, col1, col2):
    table = pd.crosstab(df[col1], df[col2])
    print(f"\nContingency Table: {col1} × {col2}")
    display(table)

categorical_contingency(df_train, "Area", "VehGas")
categorical_contingency(df_train, "VehBrand", "Region")

# ============================================
# 3. TARGET VARIABLE ANALYSIS
# ============================================

# ---------- 3.1 ClaimNb distribution ----------
plt.figure(figsize=(10, 4))
sns.histplot(df_train["ClaimNb"], bins=20)
plt.title("Distribution of ClaimNb")
plt.show()

print(df_train["ClaimNb"].value_counts().head(10))

# Zero Inflation
zero_rate = (df_train["ClaimNb"] == 0).mean()
print(f"Zero-Inflation Rate: {zero_rate:.3f}")

# ---------- 3.2 Exposure-adjusted claim rate ----------
df_train["ClaimRate"] = df_train["ClaimNb"] / df_train["Exposure"]

plt.figure(figsize=(10, 4))
sns.histplot(df_train["ClaimRate"], bins=30)
plt.title("Distribution of Exposure-adjusted Claim Rate")
plt.show()

# Check suspicious low exposure values
suspicious = df_train[df_train["Exposure"] < 0.01]
print("Suspicious low exposure rows:")
display(suspicious.head())

# ============================================
# 4. MULTIVARIATE RELATIONSHIPS
# ============================================

# ---------- 4.1 Correlation Analysis ----------
plt.figure(figsize=(10, 6))
sns.heatmap(df_train[numeric_cols].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Structure Among Predictors")
plt.show()

# ---------- 4.2 Interaction Exploration ----------
sns.lmplot(data=df_train, x="DrivAge", y="BonusMalus", hue="Area", scatter_kws={"alpha": 0.2})
plt.title("Interaction: DrivAge × BonusMalus × Area")
plt.show()

sns.boxplot(x="Region", y="ClaimNb", data=df_train)
plt.title("Interaction Example: Region × ClaimNb")
plt.xticks(rotation=45)
plt.show()

# ============================================
# 4.3 CORRELATION AFTER SCALING + SPEARMAN
# ============================================

# --- 4.3.1 Scale numeric features ---
scaler = StandardScaler()
df_scaled = df_train.copy()
df_scaled[numeric_cols] = scaler.fit_transform(df_scaled[numeric_cols])

# Pearson correlation on scaled data (same as unscaled, but okay to visualize)
plt.figure(figsize=(10, 6))
corr_pearson_scaled = df_scaled[numeric_cols].corr(method="pearson")
sns.heatmap(corr_pearson_scaled, annot=True, cmap="coolwarm")
plt.title("Pearson Correlation (Scaled Numeric Features)")
plt.show()

# --- 4.3.2 Spearman correlation (rank-based, good for non-linear monotonic) ---
plt.figure(figsize=(10, 6))
corr_spearman = df_train[numeric_cols].corr(method="spearman")
sns.heatmap(corr_spearman, annot=True, cmap="coolwarm")
plt.title("Spearman Correlation (Numeric Features)")
plt.show()

# Optional: correlation of each numeric with the target ClaimNb
print("Pearson corr with ClaimNb:")
print(df_train[numeric_cols].corrwith(df_train["ClaimNb"], method="pearson").sort_values())

print("\nSpearman corr with ClaimNb:")
print(df_train[numeric_cols].corrwith(df_train["ClaimNb"], method="spearman").sort_values())


```

```python
# ======================================================
# EXTRA EDA MODULE: ClaimRate-focused, Binned Variables,
# Calibration-style Plots, Categorical Comparisons
# ======================================================

# Ensure ClaimRate exists
df_train["ClaimRate"] = df_train["ClaimNb"] / df_train["Exposure"]

# ------------------------------------------------------
# 1. ClaimRate vs ALL numeric features
# ------------------------------------------------------
def plot_claimrate_vs_numeric(df, numeric_cols):
    for col in numeric_cols:
        plt.figure(figsize=(6,4))
        sns.scatterplot(x=df[col], y=df["ClaimRate"], alpha=0.3)
        plt.title(f"ClaimRate vs {col}")
        plt.xlabel(col)
        plt.ylabel("ClaimRate")
        plt.show()

plot_claimrate_vs_numeric(df_train, numeric_cols)

# ------------------------------------------------------
# 2. VehPower vs BINNED VehAge + PIE chart
# ------------------------------------------------------

# 2.1 Bin VehAge
df_train["VehAgeBin"] = pd.cut(
    df_train["VehAge"],
    bins=[0,5,10,15,20,100],
    labels=["0–5","5–10","10–15","15–20","20+"]
)

# 2.2 VehPower vs VehAgeBin
plt.figure(figsize=(8,5))
sns.boxplot(x="VehAgeBin", y="VehPower", data=df_train)
sns.stripplot(x="VehAgeBin", y="VehPower", data=df_train, alpha=0.1, color="black")
plt.title("VehPower vs Binned VehAge")
plt.show()

# 2.3 PIE chart
df_train["VehAgeBin"].value_counts().plot.pie(autopct="%1.1f%%", figsize=(6,6))
plt.title("Distribution of Vehicle Age (Binned)")
plt.ylabel("")
plt.show()

# ------------------------------------------------------
# 3. BINNED DrivAge vs ClaimNb (Calibration-style)
# ------------------------------------------------------

# 3.1 Bin DrivAge
df_train["DrivAgeBin"] = pd.cut(
    df_train["DrivAge"],
    bins=[18,25,35,45,55,70,120],
    labels=["18–25","25–35","35–45","45–55","55–70","70+"]
)

# 3.2 Calibration plot
calib = df_train.groupby("DrivAgeBin")["ClaimNb"].mean()
plt.figure(figsize=(8,4))
calib.plot(kind="bar")
plt.title("Average ClaimNb Across DrivAge Bins")
plt.ylabel("Mean ClaimNb")
plt.show()

# ------------------------------------------------------
# 4. Compare ClaimRate with ALL numeric features again
# ------------------------------------------------------

all_numeric = numeric_cols + ["ClaimRate"]

plt.figure(figsize=(10,6))
sns.heatmap(df_train[all_numeric].corr(method="spearman"),
            annot=True, cmap="coolwarm")
plt.title("Spearman Correlation (Including ClaimRate)")
plt.show()

plot_claimrate_vs_numeric(df_train, numeric_cols)

# ------------------------------------------------------
# 5. Exposure vs ClaimNb
# ------------------------------------------------------
plt.figure(figsize=(7,5))
sns.scatterplot(x="Exposure", y="ClaimNb", data=df_train, alpha=0.3)
plt.title("Exposure vs ClaimNb")
plt.show()

# ------------------------------------------------------
# 6. BINNED Density vs ClaimNb & ClaimRate
# ------------------------------------------------------

df_train["DensityBin"] = pd.qcut(df_train["Density"], q=5, labels=[1,2,3,4,5])

fig, ax = plt.subplots(1,2, figsize=(12,5))

sns.barplot(x="DensityBin", y="ClaimNb", data=df_train, ax=ax[0])
ax[0].set_title("ClaimNb across Density bins")

sns.barplot(x="DensityBin", y="ClaimRate", data=df_train, ax=ax[1])
ax[1].set_title("ClaimRate across Density bins")

plt.show()

# ------------------------------------------------------
# 7. Categorical bins vs ClaimNb & ClaimRate
# ------------------------------------------------------

def binned_cat_target_plots(df, cat_col):
    fig, ax = plt.subplots(1,2, figsize=(14,5))

    sns.barplot(x=cat_col, y="ClaimNb", data=df, ax=ax[0])
    ax[0].set_title(f"{cat_col} vs ClaimNb")
    ax[0].tick_params(axis="x", rotation=45)

    sns.barplot(x=cat_col, y="ClaimRate", data=df, ax=ax[1])
    ax[1].set_title(f"{cat_col} vs ClaimRate")
    ax[1].tick_params(axis="x", rotation=45)

    plt.tight_layout()
    plt.show()

for col in ["VehBrand", "Area", "Region"]:
    binned_cat_target_plots(df_train, col)

# ======================================================
# END OF CLAIMRATE EXTENDED EDA MODULE
# ======================================================
```

```python
df_train["VehAgeSimpleBin"] = pd.cut(
    df_train["VehAge"],
    bins=[-1, 3, 10, 100],
    labels=["Brand new", "Middle ground", "Old"]
)
agg = (
    df_train
    .assign(HasClaim=lambda d: d["ClaimNb"] > 0)
    .groupby(["VehPower", "VehAgeSimpleBin", "HasClaim"])
    .size()
    .reset_index(name="Count")
)
pivot = (
    agg
    .pivot_table(
        index=["VehPower", "VehAgeSimpleBin"],
        columns="HasClaim",
        values="Count",
        fill_value=0
    )
    .reset_index()
)

pivot.columns.name = None
pivot = pivot.rename(columns={False: "NoClaim", True: "Claim"})
age_colors = {
    "Brand new": "#4C72B0",
    "Middle ground": "#55A868",
    "Old": "#C44E52"
}
import matplotlib.colors as mcolors

def lighten(color, amount=0.5):
    c = np.array(mcolors.to_rgb(color))
    return tuple(c + (1 - c) * amount)
plt.figure(figsize=(14,6))
ax = plt.gca()

vehpowers = sorted(pivot["VehPower"].unique())
age_bins = pivot["VehAgeSimpleBin"].unique()

bar_width = 0.25
x = np.arange(len(vehpowers))

for i, age in enumerate(age_bins):
    sub = pivot[pivot["VehAgeSimpleBin"] == age]

    x_pos = x + (i - 1) * bar_width

    # Bottom: No claims (dark)
    ax.bar(
        x_pos,
        sub["NoClaim"],
        width=bar_width,
        color=age_colors[age],
        label=f"{age} – No claim" if i == 0 else ""
    )

    # Top: Claims (lighter same color)
    ax.bar(
        x_pos,
        sub["Claim"],
        bottom=sub["NoClaim"],
        width=bar_width,
        color=lighten(age_colors[age], 0.5),
        label=f"{age} – Claim" if i == 0 else ""
    )

# Axis formatting
ax.set_xticks(x)
ax.set_xticklabels(vehpowers)
ax.set_xlabel("Vehicle Power")
ax.set_ylabel("Frequency")
ax.set_title("VehPower × VehAge (Stacked by Claim Occurrence)")

# Clean legend (manual grouping)
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles[:6], labels[:6], ncol=2, title="Vehicle Age / Claim")

plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# BINS
# -----------------------------

df_train["VehAgeBin"] = pd.cut(
    df_train["VehAge"],
    bins=[-1, 3, 10, 100],
    labels=["Brand new", "Middle ground", "Old"]
)

df_train["DrivAgeBin"] = pd.cut(
    df_train["DrivAge"],
    bins=[18, 25, 35, 45, 55, 70, 120],
    labels=["18–25","25–35","35–45","45–55","55–70","70+"]
)

df_train["DensityBin"] = pd.qcut(
    df_train["Density"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)

df_train["VehPowerBin"] = pd.cut(
    df_train["VehPower"],
    bins=[0,5,8,12,20],
    labels=["Low","Medium","High","Very High"]
)
import matplotlib.colors as mcolors

def lighten(color, amount=0.5):
    c = np.array(mcolors.to_rgb(color))
    return tuple(c + (1 - c) * amount)

def stacked_claim_plot(
    df,
    x_col,
    hue_col,
    title,
    base_colors
):
    # Aggregate
    agg = (
        df
        .assign(HasClaim=lambda d: d["ClaimNb"] > 0)
        .groupby([x_col, hue_col, "HasClaim"])
        .size()
        .reset_index(name="Count")
    )

    pivot = (
        agg
        .pivot_table(
            index=[x_col, hue_col],
            columns="HasClaim",
            values="Count",
            fill_value=0
        )
        .reset_index()
        .rename(columns={False: "NoClaim", True: "Claim"})
    )

    x_vals = pivot[x_col].unique()
    hue_vals = pivot[hue_col].unique()

    bar_width = 0.8 / len(hue_vals)
    x = np.arange(len(x_vals))

    plt.figure(figsize=(14,6))
    ax = plt.gca()

    for i, hue in enumerate(hue_vals):
        sub = pivot[pivot[hue_col] == hue]
        xpos = x + (i - (len(hue_vals)-1)/2) * bar_width
        base_color = base_colors[hue]

        ax.bar(
            xpos,
            sub["NoClaim"],
            width=bar_width,
            color=base_color
        )
        ax.bar(
            xpos,
            sub["Claim"],
            bottom=sub["NoClaim"],
            width=bar_width,
            color=lighten(base_color, 0.5)
        )

    ax.set_xticks(x)
    ax.set_xticklabels(x_vals)
    ax.set_xlabel(x_col)
    ax.set_ylabel("Frequency")
    ax.set_title(title)

    legend_handles = [
        plt.Rectangle((0,0),1,1,color=c) for c in base_colors.values()
    ]
    ax.legend(
        legend_handles,
        base_colors.keys(),
        title=hue_col,
        ncol=len(base_colors)
    )

    plt.tight_layout()
    plt.show()
vehage_colors = {
    "Brand new": "#4C72B0",
    "Middle ground": "#55A868",
    "Old": "#C44E52"
}

density_colors = {
    "Low": "#4C72B0",
    "Medium": "#55A868",
    "High": "#C44E52",
    "Very High": "#8172B2"
}
stacked_claim_plot(
    df_train,
    x_col="VehPowerBin",
    hue_col="VehAgeBin",
    title="VehPower × VehAge (Stacked by Claim Occurrence)",
    base_colors=vehage_colors
)
stacked_claim_plot(
    df_train,
    x_col="DrivAgeBin",
    hue_col="VehAgeBin",
    title="Driver Age × Vehicle Age (Stacked by Claim Occurrence)",
    base_colors=vehage_colors
)
stacked_claim_plot(
    df_train,
    x_col="DensityBin",
    hue_col="VehAgeBin",
    title="Density × Vehicle Age (Stacked by Claim Occurrence)",
    base_colors=vehage_colors
)
stacked_claim_plot(
    df_train,
    x_col="Area",
    hue_col="VehAgeBin",
    title="Area × Vehicle Age (Stacked by Claim Occurrence)",
    base_colors=vehage_colors
)
stacked_claim_plot(
    df_train,
    x_col="Region",
    hue_col="VehAgeBin",
    title="Region × Vehicle Age (Stacked by Claim Occurrence)",
    base_colors=vehage_colors
)
```

```python
# ============================
# BIN DEFINITIONS
# ============================

df_train["VehAgeBin"] = pd.cut(
    df_train["VehAge"],
    bins=[-1, 3, 10, 100],
    labels=["Brand new", "Middle ground", "Old"]
)

df_train["DrivAgeBin"] = pd.cut(
    df_train["DrivAge"],
    bins=[18, 25, 35, 45, 55, 70, 120],
    labels=["18–25","25–35","35–45","45–55","55–70","70+"]
)

df_train["VehPowerBin"] = pd.cut(
    df_train["VehPower"],
    bins=[0,5,8,12,20],
    labels=["Low","Medium","High","Very High"]
)

df_train["DensityBin"] = pd.qcut(
    df_train["Density"],
    q=4,
    labels=["Low","Medium","High","Very High"]
)
# ============================
# VALID PLOTTING FEATURES
# ============================

plot_features = {
    "VehAgeBin": {
        "type": "binned",
        "colors": {
            "Brand new": "#4C72B0",
            "Middle ground": "#55A868",
            "Old": "#C44E52"
        }
    },
    "DrivAgeBin": {
        "type": "binned",
        "colors": {
            "18–25": "#4C72B0",
            "25–35": "#55A868",
            "35–45": "#C44E52",
            "45–55": "#8172B2",
            "55–70": "#CCB974",
            "70+": "#64B5CD"
        }
    },
    "VehPowerBin": {
        "type": "binned",
        "colors": {
            "Low": "#4C72B0",
            "Medium": "#55A868",
            "High": "#C44E52",
            "Very High": "#8172B2"
        }
    },
    "DensityBin": {
        "type": "binned",
        "colors": {
            "Low": "#4C72B0",
            "Medium": "#55A868",
            "High": "#C44E52",
            "Very High": "#8172B2"
        }
    },
    "Area": {"type": "categorical"},
    "Region": {"type": "categorical"},
    "VehBrand": {"type": "categorical"}
}
import matplotlib.colors as mcolors

def lighten(color, amount=0.5):
    c = np.array(mcolors.to_rgb(color))
    return tuple(c + (1 - c) * amount)

def stacked_claim_plot(df, x_col, hue_col, title, base_colors):
    agg = (
        df
        .assign(HasClaim=lambda d: d["ClaimNb"] > 0)
        .groupby([x_col, hue_col, "HasClaim"])
        .size()
        .reset_index(name="Count")
    )

    pivot = (
        agg
        .pivot_table(
            index=[x_col, hue_col],
            columns="HasClaim",
            values="Count",
            fill_value=0
        )
        .reset_index()
        .rename(columns={False: "NoClaim", True: "Claim"})
    )

    x_vals = pivot[x_col].unique()
    hue_vals = pivot[hue_col].unique()

    bar_width = 0.8 / len(hue_vals)
    x = np.arange(len(x_vals))

    plt.figure(figsize=(14,6))
    ax = plt.gca()

    for i, hue in enumerate(hue_vals):
        sub = pivot[pivot[hue_col] == hue]
        xpos = x + (i - (len(hue_vals)-1)/2) * bar_width
        base_color = base_colors.get(hue, "#4C72B0")

        ax.bar(xpos, sub["NoClaim"], width=bar_width, color=base_color)
        ax.bar(
            xpos,
            sub["Claim"],
            bottom=sub["NoClaim"],
            width=bar_width,
            color=lighten(base_color)
        )

    ax.set_xticks(x)
    ax.set_xticklabels(x_vals, rotation=45)
    ax.set_xlabel(x_col)
    ax.set_ylabel("Frequency")
    ax.set_title(title)

    plt.tight_layout()
    plt.show()
# ============================
# AUTO-GENERATE VALID PAIRS
# ============================
import itertools

MAX_LEVELS = 10

features = list(plot_features.keys())

for x_col, hue_col in itertools.combinations(features, 2):
    # Skip very high-cardinality combinations
    if df_train[x_col].nunique() > MAX_LEVELS:
        continue
    if df_train[hue_col].nunique() > MAX_LEVELS:
        continue

    # Pick colors from hue feature if available
    base_colors = plot_features.get(hue_col, {}).get("colors", {})

    title = f"{x_col} × {hue_col} (Stacked by Claim Occurrence)"

    stacked_claim_plot(
        df_train,
        x_col=x_col,
        hue_col=hue_col,
        title=title,
        base_colors=base_colors
    )
```
